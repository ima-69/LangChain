### Sentence Transformers on HugginFace

Hugging face sentence-transformers is a python framework for state-of-the-art sentence, text and image embeddings. One of the embedding models is used in the HuggingFaceEmbeding class. We have also added an alias for SentenceTransformerEmbeddings for users who are more familier with directly using that package.

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
text = "LangChain is a framework for developing applications powered by language models."
query_result = embeddings.embed_query(text)
query_result
```
