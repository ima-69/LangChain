### Embedding Techniques

converting text into vectors



```python
import os
from dotenv import load_dotenv
load_dotenv() #load all environment variables from a .env file
```

```python
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import OpenAIEmbeddings
embeddings = OpenAIEmbeddings(model="text-embedding-3-large")
embeddings
```

```python
text = "LangChain is a framework for developing applications powered by language models."
embedding_vector = embeddings.embed_query(text)
print(embedding_vector)
```

```python
len(embedding_vector)  # Check the length of the embedding vector
```

```python
embeddings_1024 = OpenAIEmbeddings(model="text-embedding-3-large", dimensions=1024)
embedding_vector_1024 = embeddings_1024.embed_query(text)
print(len(embedding_vector_1024))  # Should print 1024
```

```python
len(embedding_vector_1024)  # Check the length of the embedding vector
```

```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader("speech.txt")
documents = loader.load()
documents
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
final_documents = text_splitter.split_documents(documents)
final_documents
```

```python
## Vector Embeddings and Vector StoreDB
from langchain_community.vectorstores import Chroma
db = Chroma.from_documents(final_documents, embeddings_1024)
db
```

```python
## Retrieve the results from querying the vector store

query = "It will be all the easier for us to conduct ourselves as belligerents in a high spirit of right and fairness"
results = db.similarity_search(query)
results
```
