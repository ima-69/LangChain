### Ollama

Ollama support embedding models, making it possible to build retrieval augmented generation(RAG) applications that combine text prompts with existing documents or other data.

```python
from langchain_community.embeddings import OllamaEmbeddings
```

```python
embeddings = (
    OllamaEmbeddings(model="gemma2:2b")  ## By default, it uses "gemma2"
)
```

```python
embeddings
```

```python
r1=embeddings.embed_documents(
    [
        "Alpha is the first letter of the Greek alphabet.",
        "Beta is the second letter of the Greek alphabet.",
    ]
)
```

```python
r1
```

```python
len(r1[1])
```

```python
embeddings.embed_query("What is the second letter of the Greek alphabet?")
```

```python
### Other Embedding Models

embeddings = OllamaEmbeddings(model="mxbai-embed-large")
text = "LangChain is a framework for developing applications powered by language models."
query_result = embeddings.embed_query(text)
query_result

```
